# Importing Libraries and Setting Selenium Options

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.action_chains import ActionChains
import time
import pandas as pd
```

```python
options = Options()
options.add_argument("--disable-blink-features=AutomationControlled")
options.add_experimental_option("excludeSwitches", ["enable-automation"])
options.add_experimental_option('useAutomationExtension', False)
options.add_argument("--start-minimized")
options.add_argument("--window-size=1920,1080")
options.add_argument("--disable-gpu")
options.add_argument("--no-sandbox")
options.add_argument("--disable-dev-shm-usage")
```

# Scrapping Data

```python
driver = webdriver.Chrome(options=options)
number_of_pages = 3
list_jobs = []

print(f"Scraping {number_of_pages} pages...")
for page in range(1, number_of_pages + 1):
    url = f"https://www.naukrigulf.com/data-engineer-jobs-{page}"
    driver.get(url)
    time.sleep(3)
    jobs = driver.find_elements(By.CSS_SELECTOR, ".ng-box.srp-tuple")
    for job in jobs:
        job_title = job.find_element(By.CSS_SELECTOR, "p.designation-title").text
        try:
            job_company = job.find_element(By.CSS_SELECTOR, "a.info-org").text
        except:
            job_company = job.find_element(By.CSS_SELECTOR, "p.info-org").text
        job_location = job.find_element(By.CSS_SELECTOR, "li.info-loc").text
        job_experience = job.find_element(By.CSS_SELECTOR, "li.info-exp").text
        job_description = job.find_element(By.CSS_SELECTOR, "p.description").text
        list_jobs.append({
            "Job Title": job_title,
            "Company Name": job_company,
            "Job Location": job_location,
            "Required Experience": job_experience,
            "Job Description": job_description
        })    

driver.quit()

print(f"Found {len(list_jobs)} jobs in {number_of_pages} pages")
```

# Cleaning and Saving Data

```python
df = pd.DataFrame(list_jobs)

# Drop duplicates
df = df.drop_duplicates()

print(f"Total jobs after cleaning: {len(df)}")

# Save the cleaned data to a CSV file
df.to_csv('data_engineer_jobs.csv', index=False)
print("Data saved to data_engineer_jobs.csv")


# Note: The text contains commas and pipe characters, so using them as 
# separators causes some fields to be wrapped in double quotation marks in some editors.
# To avoid this problem, use a tab separator:
# df.to_csv('data_engineer_jobs.tsv', index=False, sep='\t')
# print("Data saved to data_engineer_jobs.tsv")
```
